```python
#1-10
for i in range(1, 11):
    print(i)
print("\n")
#evens
for i in range(1, 21, 2):
    print(i)
print("\n")
#counting backwards
for i in range(10, 0, -1):
  print(i)
print("\n")
#multiplcation table
num=int(input("Enter a number for the multiplication table: "))
for i in range(1, 11):
    print(f"{num} x {i} = {num*i}")
print("\n")
#summation
total=0
for i in range(1,11):
  total += i
  print(total)
print("\n")
#test scores
total = 0
num_scores = 5
print("Please enter 5 grades:")
for i in range(1, num_scores + 1):
    score = float(input(f"Enter score: "))
    total += score

print(f"Total scores: {total}")
print(f"Average score is {total / num_scores:.2f}")
```
