```python
#3 funcs
title = input("Enter a title: ")
def display_title(title):
  print(f"Title: {title}")

short_message = input("Enter a short message: ")
def display_short_message(short_message):
  print(f"Short Message: {short_message}")

closing_message = input("Enter a closing message: ")
def display_closing_message(closing_message):
  print(f"Closing Message: {closing_message}")

display_title(title)
display_short_message(short_message)
display_closing_message(closing_message)
```
