```python
#list of food
foods = ["pizza", "tacos", "pasta", "salad", "burgers"]
print(foods[1])
print(foods[3])
print(foods[-1])
```

```python
#change item
colors = ["red", "blue", "green", "yellow"]
print(colors)
colors[0] = "purple"
print(colors)
```

```python
#add items to a list
courses = ["COP 1000", "ENC 1101"]
courses.append("CGS 1060")
print(courses)
courses.insert(0, "MAC 1147")
print(courses)
```

```python
#remove items
shopping = ["milk", "bread", "eggs", "apples", "coffee"]
print(shopping)
shopping.remove("eggs")
print(shopping)
print(len(shopping))
```

```python
#sort a list
names = ["Aiden", "Jaden", "Haden", "Caden", "Aden"]
names.sort()
print(names)
names.reverse()
print(names)
```

```python
#work with numbers in a list
scores = [88, 94, 76, 91, 85]
print(max(scores))
print(min(scores))
print(sum(scores))
print(len(scores))
average = sum(scores) / len(scores)
print(f"{average:.2f}")
```

```python
#simple student list
students = []
for i in range(3):
  name = input(f"Enter name {i+1}: ")
  students.append(name)
print(students)
print(len(students))
students.sort()
print(students)
```
