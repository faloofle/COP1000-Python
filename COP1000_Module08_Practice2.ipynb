```python
#import the csv module
import csv
display("csv tools are ready")
```

```python
#create a csv file
with open('students.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Major', 'Credits'])
    writer.writerows([
        ['Alex', 'Cybersecurity', 24],
        ['Jordan', 'Information Technology', 36],
        ['Taylor', 'Computer Science', 42]
    ])
```

```python
#read a csv file
with open('students.csv', mode='r', encoding='utf-8') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
#display selected data
with open('students.csv', mode='r', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row['Name'], row['Major'])
```

```python
#add a new record
newname=input("enter a name: ")
newmajor=input("enter a major: ")
newcredits=input("enter credits: ")
with open('students.csv', mode='a', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow([newname, newmajor, newcredits])
with open('students.csv', mode='r', encoding='utf-8') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
#calculate data from csv
import csv

with open('students.csv', mode='r', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    credits = [int(row['Credits']) for row in reader]

record_count = len(credits)
total_credits = sum(credits)
average_credits = total_credits / record_count if record_count > 0 else 0.0

print(f"Student Records: {record_count}")
print(f"Total Credits: {total_credits}")
print(f"Average Credits: {average_credits:.2f}")
```

```python
#create my own csv dataset
import csv
with open('products.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(['Product', 'Price', 'Quantity'])
    writer.writerows([
        ['Notebook', 3.49, 4],
        ['Pen', 1.25, 10],
        ['Folder', 2.00, 5]
    ])

with open('products.csv', mode='r', encoding='utf-8') as file:
    reader = csv.reader(file)
    print("--- products.csv Contents ---")
    for row in reader:
        print(row)
```
