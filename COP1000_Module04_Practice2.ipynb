```python
#number 1=10
number = 1
while number <= 10:
    print(number)
    number += 1
print("\n")
#countdown
number = 5
while number <= 5 and number >= 1:
  print(number)
  number -= 1
print("Done!" ,"\n")
#continuing until stopping
response = " "
while response.lower() != "no":
    response = input("Would you like to continue? Enter yes or no: ")
    if response.lower() == "yes":
        print("Continuing...")
    elif response.lower() == "no":
        print("Program finished.")
    else:
      print("Invalid")
#password
password="python123"
response=" "
while response != "python123":
  response = input("Enter Password: ")
  if response!=password:
    print("Incorrect Password")
  else:
    print("Correct Password")
print("\n")
#number range
response = 0
while response < 1 or response > 10:
  response = int(input("Enter a number between 1 and 10: "))
  if response >= 1 and response <= 10:
    print(f"You entered a valid number {response}")
  else:
    print(f"You entered an invalid number {response}")
print("\n")
#running total until zero
total = 0
number_input = int(input("Enter a number or 0 to stop: "))
while number_input != 0:
  total += number_input
  number_input = int(input("Enter a number or 0 to stop: "))
print(f"Total: {total}")
print("\n")
#limited attempts
attempts = 0
max_attempts = 3
correct_word = "python"
while attempts < max_attempts:
    response = input("Enter the word ")
    if response.lower() == correct_word:
        print("Correct!")
        attempts = max_attempts
    else:
        print("Incorrect word.")
        attempts += 1
        if attempts == max_attempts:
            print("No attempts remaining")
```
