```python
#create a calculation method
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width
rectangle1 = Rectangle(length=5, width=3)
rectangle2 = Rectangle(length=4, width=6)
print("Rectangle 1 Area:", rectangle1.calculate_area())
print("Rectangle 2 Area:", rectangle2.calculate_area())
```

```python
#create a method that changes an attribute
class BankAccount:
    def __init__(self, account_name, balance):
        self.account_name = account_name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
account1 = BankAccount(account_name="Aiden", balance=1000)
print("Initial Balance:", account1.balance)
account1.deposit(amount=500)
print("Updated Balance:", account1.balance)
```

```python
# add another behavior
class BankAccount:
    def __init__(self, account_name, balance):
        self.account_name = account_name
        self.balance = balance

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self.balance}")
        else:
            print("Insufficient funds")

account1 = BankAccount(account_name="Aiden", balance=1500)
account1.withdraw(amount=200)
```

```python
#model a book
class Book:
    def __init__(self, title, author, availability):
        self.title = title
        self.author = author
        self.availability = availability

    def check_out(self):
      self.availability = False
      print(f"{self.title} has been checked out.")
    def check_in(self):
      self.availability = True
      print(f"{self.title} has been checked in.")
book = Book(title="The Great Gatsby", author="F. Scott Fitzgerald", availability=True)
book.check_out()
book.check_in()
```

```python
#multiple objects with methods
class Book:
    def __init__(self, title, author, availability):
        self.title = title
        self.author = author
        self.availability = availability

    def check_out(self):
        self.availability = False
        print(f"{self.title} has been checked out.")

    def check_in(self):
        self.availability = True
        print(f"{self.title} has been checked in.")

book1 = Book(title = "The Hunger Games", author = "Suzanne Collins", availability = True)
book2 = Book(title = "Diary of a Wimpy Kid", author = "Jeff Kinney", availability = True)
book3 = Book(title = "Leaves of Grass", author = "Walt Whitman", availability = True)

for book in [book1, book2, book3]:
    print(f"Title: {book.title}\nAuthor: {book.author}\nAvailability: {book.availability}\n")
book1.check_out()
book2.check_in()
book3.check_out()
#
```

```python
#design your own class
class Character:
    def __init__(self, name, species, morality):
        self.name = name
        self.species = species
        self.morality = morality
    def display_info(self):
      print(f"Name: {self.name}\nSpecies: {self.species}\nMorality: {self.morality}\n")
    def change_species(self, new_species):
      self.species = new_species
      print(f"{self.name}'s species has been changed to {self.species}.\n")

character1 = Character(name="Aiden", species="Human", morality="Good")
character1.display_info()
character1.change_species("Alien")
character1.display_info()
```
