```python
#import the math module
import math
display(math.pi)
display(math.sqrt(81))
display(math.pow(5,2))
```

```python
#calculate the area of a circle
radius = input("Enter the radius of the circle: ")
radius = float(radius)
area = math.pi * (radius ** 2)
print(f"The area of the circle is: {area:.2f}")
```

```python
#use the random module
import random
random.randint(1,10)
```

```python
#simulate a coin flip
outcomes = ['Heads', 'Tails']

result = random.choice(outcomes)
display(result)
```

```python
#simulate a six sided die
die = [1,2,3,4,5,6]

for i in range(5):
  result = random.choice(die)
  display(f"Roll {i+1} : {result}")

```

```python
#import specific functions
from math import sqrt
display(sqrt(25))
display(sqrt(100))
display(sqrt(144))
```

importing the whole math module allows you to use whatever is inside the whole module whereas just importing a singular function only allows you to use the single function, it saves having to get a whole module just for a single function.
