```python
#greetings
firstname=input("Enter your first name: ")
print("Hello " + firstname)
city=input("Enter your city: ")
favorite_food=input("Enter your favorite food: ")
print(f"{firstname} lives in {city} and their favorite food is {favorite_food}. \n")

#age
age=input("Enter your age: ")
age=int(age)
print(f"You are currently {age} and will be {age+1} next year \n")

#adding 2 numbers
number1=input("Enter a number: ")
number2=input("Enter another number: ")
number1=int(number1)
number2=int(number2)
print(f"The sum of both numbers are {number1+number2} \n")

#purchase
item=input("Enter the name of the item: ")
price=input("Enter the price of the item: ")
quantity=input("Enter the quantity of the item: ")
price=float(price)
quantity=int(quantity)
print(f"Purchase Summary: \n Item: {item} \n Price: ${price} \n Quantity: {quantity} \n Total Cost: ${price*quantity:.2f} \n")

#gross pay
hoursworked=input("Enter the amount of hours you worked: ")
hourlyrate=input("Enter your hourly rate: ")
hoursworked=float(hoursworked)
hourlyrate=float(hourlyrate)
print(f"Hours Worked: {hoursworked} \nHourly Rate: ${hourlyrate} \nGross Pay : ${hourlyrate*hoursworked:.2f}")

```
