```python
#identify a runtime error
number = 10
try:
    result = number / 0
    print(result)
except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

```python
#debug a loop
count = 1

while count <= 5:
    print(count)
    count=count+1
```

the original problem was that the count did not change inside the loop. meaning it is constantly one, but since one is less than five it is an infinite loop so it will print one forever. adding a count=count+1 makes it so that it goes up every loop, and then it will reach the limits of the argument in the while function


```python
def calculate_discount(price, discount_percent):
    new_price = price * (1-(discount_percent / 100))
    return new_price

price = float(input("enter price: "))
discount_percent = float(input("enter discount: "))

new_price = calculate_discount(price, discount_percent)
print(f"the original price was: {price:.2f}, your new price is: {new_price:.2f}")
```
