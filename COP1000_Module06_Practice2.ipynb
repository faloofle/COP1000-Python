```python
#loop through a list
animals = ["dog", "cat", "rabbit", "horse", "turtle"]
for animal in animals:
    print(animal)
```

```python
#number the items
tasks = ["Study Python", "Complete homework", "Read chapter", "Review notes"]
for i, task in enumerate(tasks, start=1):
    print(f"{i}. {task}")
```

```python
#search a list
usernames = ["alex", "jamie", "morgan", "taylor"]
if input("Enter a username: ".lower()) in usernames:
    print("Username Found")
else:
    print("Username Not Found")
```

```python
#filter values
temperatures = [45, 72, 88, 61, 93, 55, 79]
count = 0
for temp in temperatures:
    if temp >= 70:
        print(temp)
        count += 1
print(f"Total temperatures that reached 70+ degrees: {count}")
```

```python
#calculate an average from user data
scores = []
for i in range(5):
    score = int(input(f"Enter score {i+1} of 5: "))
    scores.append(score)

average = sum(scores) / len(scores)
print(f"The average score is: {average:.2f}")
```

```python
#create a tuple
weekend = ("Saturday", "Sunday")
print(weekend)
print(weekend[0])
print(weekend[1])
print(len(weekend))
for weekend_day in weekend:
    print(weekend_day)
```

```python
#lists vs tuples
colors_list = ["red", "green", "blue"]
print(colors_list)
colors_tuple = ("red", "green", "blue")
colors_list[0] = "purple"
print(colors_list)
```

tuples are immutable, meaning what is in the list cannot be modified, so the normal changing of elements in a normal list will not work with tuples.
