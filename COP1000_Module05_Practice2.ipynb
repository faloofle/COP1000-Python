```python
#add 2 numb
def add_numbers(number1, number2):
  print(f"Number 1: {number1}\nNumber 2: {number2}\nSum: {int(number1)+int(number2)}")

number1 = input("enter a number: ")
number2 = input("enter a number: ")
add_numbers(number1, number2)
```

```python
#rectangel
def calculate_area(length, width):
  print(f"Length: {length}\nWidth: {width}\nArea: {int(length)*int(width)}")

length = input("enter a length: ")
width = input("enter a width: ")
calculate_area(length, width)
```

```python
#fahren to cels
def fahrenheit_to_celsius(fahrenheit):
  print(f"Fahrenheit: {fahrenheit}\nCelsius: {(int(fahrenheit)-32)*5/9}")

fahrenheit = input("enter a fahrenheit: ")
fahrenheit_to_celsius(fahrenheit)
```

```python
#letter grade
def letter_grade(score):
  score = int(score)
  if score >= 90:
    return "A"
  elif score >= 80:
    return "B"
  elif score >= 70:
    return "C"
  elif score >= 60:
    return "D"
  else:
    return "F"

score = input("Input Score: ")
print(letter_grade(score))
```

```python
#sep calc from out
def calculate_total(price, quantity):
  return (price * quantity)
def display_total(item, total,):
  print(f"{item}, ${total:.2f}")

item=input("Enter the item name: ")
price=input("Enter the item price: ")
quantity=input("Enter the item quantity: ")
price=float(price)
quantity=int(quantity)
total=calculate_total(price, quantity)
display_total(item, total)
```

```python
#multi func programs
def total_damage(raw, strength, cc, cs):
  raw=float(raw)
  strength=float(strength)
  cc=float(cc)
  cs=float(cs)
  return (raw * (strength/strength*1.8) + ((cc/100) * cs))
  print("Damage: ", total_damage)
raw=input("Input Raw damage: ")
strength=input("Input Strength: ")
cc=input("Input Defense Penetration Percent (0-100): ")
cs=input("Input Defense Penetration Damage: ")
total_damage(raw, strength, cc, cs)
```
