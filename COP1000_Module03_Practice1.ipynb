```python
#variable
number1=18
number2=5

#calculations and display
print("Addition", number1,"+",number2,"=",number1+number2)
print("Subtraction", number1,"-",number2,"=",number1-number2)
print("Multiplication", number1,"*",number2,"=",number1*number2)
print("Division", number1,"/",number2,"=",number1/number2)
print("Floor Division", number1,"//",number2,"=",number1//number2)
print("Modulus", number1,"%",number2,"=",number1%number2)
print("Exponentation", number1,"**",number2,"=",number1**number2, "\n")

#comparing
age=21
print("Is age equal to 21?", age==21)
print("Is age not equal to 18?", age!=18)
print("Is age greater than 18?", age>18)
print("Is age less than 30?", age<30)
print("Is age greater than or equal to 21?", age>=21)
print("Is age less than or equal to 20?", age<=20)

```

```python
print(10 > 5)
print(8 == 8)
print(7 != 7)
print(4 >= 9)
print(15 < 20, "\n")

#boolean operators
age=22
has_id= True
print(age>=18 and has_id)
print(age<18 or has_id)
print(not has_id, "\n")

#combine comparisions
temperature=75
print(temperature>70)
print(temperature<90)
print(70<temperature<90)
print(temperature < 60 or temperature > 85,"\n")

#eligibility check
age=19
credits_completed=30
print("Age Requirement Met:",age>=18)
print("Credit Requirement Met:",credits_completed>=24)
print("All Requirements Met:",age>=18 and credits_completed>=24)
```
