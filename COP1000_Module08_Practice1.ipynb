```python
#create a simple notes file
note1 = input("Enter your first note: ")
with open("notes.txt", "w") as file:
    file.write(note1)
note2 = input("Enter your second note: ")
with open("notes.txt", "a") as file:
    file.write("\n" + note2)
note3 = input("Enter your third note: ")
with open("notes.txt", "a") as file:
    file.write("\n" + note3)
with open("notes.txt", "r") as file:
    print(file.read())
```
