```python
#create a dictionary
student = {
    "name": "Jordan",
    "major": "Cybersecurity",
    "credits": 30
}
print(student)
print(student["name"])
print(student["major"])
print(student["credits"])
```

```python
#add a new key value pair
student["gpa"] = 3.8

print(student)
print(student["gpa"])
```

```python
#update a dictionary value
course = {
    "course": "COP 1000",
    "room": "A101",
    "students": 24
}
print(course)

course["room"] = "B205"
course["students"] = 26

print(course)
```

```python
#remove data
profile = {
    "name": "Taylor",
    "city": "Jacksonville",
    "age": 21,
    "favorite_color": "blue"
}
print(profile)
del profile["favorite_color"]
print(profile)
print(len(profile))
```

```python
#check for a key
inventory = {
    "notebooks": 12,
    "pens": 30,
    "folders": 8
}
if "pens" in inventory:
    print("Yes, pens is a key in the inventory.")
else:
    print("No, pens is not a key in the inventory.")
if "markers" in inventory:
    print("Yes, markers is a key in the inventory.")
else:
    print("No, markers is not a key in the inventory.")
```

```python
#loop through a dictionary
prices = {
    "Notebook": 3.49,
    "Pen": 1.25,
    "Folder": 2.00,
    "Backpack": 24.99
}
for key, value in prices.items():
    print(f"{key}: ${value:.2f}")
```

```python
#build a simple contact record
contact = {}
contact["name"] = input("Name: ")
contact["phone"] = input("Phone: ")
contact["email"] = input("Email: ")
contact["city"] = input("City: ")
print("\n")
for key, value in contact.items():
    print(f"{key}: {value}")
```
