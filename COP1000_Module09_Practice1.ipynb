```python
#handle a valueerror
try:
    age = int(input("Enter your age: "))
    print(f"You are {age} years old.")
except ValueError:
    print("Please enter a valid whole number.")
```

```python
#protect a division calculation
try:
    numerator = int(input("Enter the numerator: "))
    denominator = int(input("Enter the denominator: "))
    quotient = numerator / denominator
    print(f"The result is {quotient}")
except ValueError:
    print("Please enter valid whole numbers.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

```python
#validate a grade
try:
    grade = int(input("Enter your grade: "))
    if grade < 0 or grade > 100:
        print("The grade is invalid. It must be between 0 and 100.")
    else:
        print(f"The grade {grade} is valid.")
except ValueError:
    print("Please enter a valid whole number.")
```

```python
#keep asking until input is valid
while True:
    try:
        age = int(input("Enter your age: "))
        if age < 0:
            print("Age must be a positive number. Please try again.")
            continue
        print(f"{age} is a valid age.")
        break
    except ValueError:
        print("Please enter a valid whole number.")
```

```python
#handle a missing file
try:
    with open("grades.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("The requested file could not be found.")
```

```python
#safe average calculator
try:
    score1 = float(input("Enter the first test score: "))
    score2 = float(input("Enter the second test score: "))
    score3 = float(input("Enter the third test score: "))
    average = (score1 + score2 + score3) / 3
    print(f"The average score is {average:.2f}")
except ValueError:
    print("Please enter a valid numeric value for each test score.")
```
