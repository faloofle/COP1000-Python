```python
#create a set
colors = {"red", "blue", "green", "yellow"}
len(colors)
```

```python
#add remove values
courses = {"COP 1000", "ENC 1101", "MAT 1033"}
courses.add("SLS 1101")
courses.remove("MAT 1033")
print(courses)
```

```python
#remove duplicate values
names = ["Alex", "Jordan", "Alex", "Taylor", "Jordan", "Morgan"]
unique_names = set(names)
print(unique_names)
print(len(names))
print(len(unique_names))
```

```python
#test membership
languages = {"python", "java", "c++", "javascript"}
inputlang = input("Enter a language: ").lower()
if inputlang in languages:
    print(f"{inputlang} is in the set")
else:
    print(f"{inputlang} is not in the set")
```

```python
#find common value
student_a = {"Python", "Gaming", "Music", "Travel"}
student_b = {"Python", "Music", "Photography", "Sports"}


common_interests = student_a & student_b
print("Common interests:", common_interests)
```

```python
#find all unique values
student_a = {"Python", "Gaming", "Music", "Travel"}
student_b = {"Python", "Music", "Photography", "Sports"}


unique_interests = student_a ^ student_b
print("Unique interests:", unique_interests)
```

```python
#compare course enrollment
python_students = {"Alex", "Jordan", "Taylor", "Morgan"}
web_students = {"Jordan", "Morgan", "Casey", "Riley"}

both_courses = python_students & web_students
print("Students in both courses:", both_courses)
all_students = python_students | web_students
print("All students:", all_students)
python_only = python_students - web_students
print("Python students:", python_only)
```
