```python
# create multiple objects
class Student:
    def __init__(self, name, major):
        self.name = name
        self.major = major

student1 = Student("Jordan", "IT")
student2 = Student("Alex", "CS")
student3 = Student("Taylor", "Graphic Design")

print(f"Name: {student1.name}\nMajor: {student1.major}\n")
print(f"Name: {student2.name}\nMajor: {student2.major}\n")
print(f"Name: {student3.name}\nMajor: {student3.major}\n")
```

```python
#create class with __init__
class Book:
  def __init__(self, title, author):
    self.title = title
    self.author = author

book1 = Book("Harry Potter", "J.K. Rowling")
book2 = Book("Lord of the Rings", "J.R.R. Tolkien")

print(f"Book 1: {book1.title} by {book1.author}")
print(f"Book 2: {book2.title} by {book2.author}")
```

```python
#add more attributes
class Book:
  def __init__(self, title, author, year, availability):
    self.title = title
    self.author = author
    self.year = year
    self.availability = availability

book1 = Book("Harry Potter", "J.K. Rowling", 1997, True)
book2 = Book("Lord of the Rings", "J.R.R. Tolkien", 1954, False)

print(f"Book 1: {book1.title} by {book1.author} ({book1.year}) - Available: {book1.availability}")
print(f"Book 2: {book2.title} by {book2.author} ({book2.year}) - Available: {book2.availability}")
```

```python
#update an object attribute
class Product:
    def __init__(self, name, price, quantity):
      self.name = name
      self.price = price
      self.quantity = quantity

product1 = Product("Apple", 1, 10)

print(f"Quantity: {product1.quantity}")

product1.quantity = 20

print(f"Quantity: {product1.quantity }")
```

```python
#use objects in a list
class Book:
  def __init__(self, title, author):
    self.title = title
    self.author = author

book1 = Book("The Hobbit", "J.R.R. Tolkien")
book2 = Book("To Kill a Mockingbird", "Harper Lee")
book3 = Book("1984", "George Orwell")

books = [book1, book2, book3]

for book in books:
    print(f"Title: {book.title} | Author: {book.author}")
```

```python
#create your own class
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

vehicle1 = Vehicle("Toyota", "Camry", 2022)
vehicle2 = Vehicle("Volkswagen", "Golf GTI", 2012)

print(f"Vehicle 1: {vehicle1.year} {vehicle1.make} {vehicle1.model}")
print(f"Vehicle 2: {vehicle2.year} {vehicle2.make} {vehicle2.model}")
```
